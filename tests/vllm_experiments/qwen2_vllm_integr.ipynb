```python
import time
from vllm import LLM, SamplingParams
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

PROMPT = prompt = "What joke Euler would tell to Turing?"
REPEAT = 10  # Number of runs per backend
MAX_TOKENS = 128
MODEL = "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"  # Replace with actual Qwen2 repo if different
torch_dtype = torch.bfloat16  # You can use float16 or bfloat16

def benchmark_vllm():
    llm = LLM(
        model=MODEL,
        tokenizer=MODEL,
        trust_remote_code=True,
    )
    sampling_params = SamplingParams(max_tokens=MAX_TOKENS)
    print("Running warmup for vLLM ...")
    llm.generate([PROMPT], sampling_params)  # Warmup
    times = []
    n_tokens = []
    for _ in range(REPEAT):
        start = time.time()
        outputs = llm.generate([PROMPT], sampling_params)
        latency = time.time() - start
        gen_text = outputs[0].outputs[0].text
        n_gen = len(outputs[0].outputs[0].token_ids)
        times.append(latency)
        n_tokens.append(n_gen)
    print(f"vLLM: Avg time: {sum(times)/REPEAT:.3f}s, Avg tokens/s: {sum(n_tokens)/sum(times):.2f}")

def benchmark_torch():
    tokenizer = AutoTokenizer.from_pretrained(MODEL, trust_remote_code=True)
    model = AutoModelForCausalLM.from_pretrained(MODEL, trust_remote_code=True).to('cuda', dtype=torch_dtype)
    print("Running warmup for HuggingFace Transformers ...") 
    inputs = tokenizer(PROMPT, return_tensors="pt").to("cuda")
    _ = model.generate(**inputs, max_new_tokens=MAX_TOKENS)
    times = []
    n_tokens = []
    for _ in range(REPEAT):
        inputs = tokenizer(PROMPT, return_tensors="pt").to("cuda")
        torch.cuda.synchronize()
        start = time.time()
        outputs = model.generate(**inputs, max_new_tokens=MAX_TOKENS)
        torch.cuda.synchronize()
        latency = time.time() - start
        gen = tokenizer.decode(outputs[0][inputs['input_ids'].shape[-1]:], skip_special_tokens=True)
        n_gen = outputs.shape[-1] - inputs['input_ids'].shape[-1]
        times.append(latency)
        n_tokens.append(n_gen)
    print(f"PyTorch: Avg time: {sum(times)/REPEAT:.3f}s, Avg tokens/s: {sum(n_tokens)/sum(times):.2f}")
```
